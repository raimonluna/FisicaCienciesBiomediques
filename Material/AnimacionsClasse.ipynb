```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

frames = 200
w      = 0.05
v      = 1

fig, ax = plt.subplots(2, 1, figsize = (10,6))
plt.rc('font', size = 16)
plt.close()

x = np.linspace(0, 1, 100)
t = np.linspace(0, 2, frames)
p = np.linspace(0, 2 * np.pi, 10)

ax[0].set_ylim(-1.1, 1.1)
ax[1].set_ylim(-1.1, 1.1)
ax[1].set_xlim(-0.05, 1.0725)
ax[0].set_xlabel('x')
ax[0].set_ylabel('y')
ax[1].set_xlabel('x')
ax[1].set_ylabel('y')
ax[0].grid()
ax[1].grid()

ln0, = ax[0].plot(x, 0.5 * np.exp(- (x)**2 / w**2), '-b', lw = 3)
ln1, = ax[1].plot(x, 0.5 * np.exp(- (x)**2 / w**2), '-b', lw = 3)

ax[0].plot([1, 1], [-2, 2], lw = 3, color = 'brown')
ax[1].plot([1.02, 1.02], [-2, 2], lw = 3, color = 'brown')
ax[0].plot([1], [0], 'ob')
hoop1, = ax[1].plot(1.02 + 0.02 * np.cos(p)[:5], 0.05 * np.sin(p)[:5], 'b', lw = 3, zorder = 0)
hoop2, = ax[1].plot(1.02 + 0.02 * np.cos(p)[5:], 0.05 * np.sin(p)[5:], 'b', lw = 3, zorder = 2)

def update(i):
    height = 0.5 * np.exp(- (1 - v * t[i])**2 / w**2) + 0.5 * np.exp(- (1 - 2 + v * t[i])**2 / w**2)
    ln0.set_data(x, 0.5 * np.exp(- (x - v * t[i])**2 / w**2) - 0.5 * np.exp(- (x - 2 + v * t[i])**2 / w**2))
    ln1.set_data(x, 0.5 * np.exp(- (x - v * t[i])**2 / w**2) + 0.5 * np.exp(- (x - 2 + v * t[i])**2 / w**2) )
    hoop1.set_data(1.02 + 0.02 * np.cos(p)[:5], 0.05 * np.sin(p)[:5] + height)
    hoop2.set_data(1.02 + 0.02 * np.cos(p)[5:], 0.05 * np.sin(p)[5:] + height)

plt.tight_layout()
animation_fig = animation.FuncAnimation(fig, update, frames = frames, interval = 50)
#HTML(animation_fig.to_jshtml())
animation_fig.save("ANIM01_WaveReflection.mp4", dpi = 200)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

frames = 200
w      = 6 * np.pi
v      = 1
k      = w / v

fig, ax = plt.subplots(2, 1, figsize = (10,6))
plt.rc('font', size = 16)
plt.close()

x = np.linspace(0, 1, 100)
t = np.linspace(0, 2, frames)

ax[0].set_ylim(-1.1, 1.1)
ax[1].set_ylim(-1.1, 1.1)
ax[0].set_xlabel('x')
ax[0].set_ylabel('y')
ax[1].set_xlabel('x')
ax[1].set_ylabel('y')
ax[0].grid()
ax[1].grid()

ln0, = ax[0].plot(x, np.cos(k *x - w * t[0]), '-b', lw = 3)
ln1, = ax[1].plot(x, np.cos(k *x + w * t[0]), '-b', lw = 3)

def update(i):
    ln0.set_data(x, np.cos(k *x - w * t[i]) )
    ln1.set_data(x, np.cos(k *x + w * t[i]) )

plt.tight_layout()
animation_fig = animation.FuncAnimation(fig, update, frames = frames, interval = 50)
#HTML(animation_fig.to_jshtml())
animation_fig.save("ANIM02_HarmonicWaves.mp4", dpi = 200)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

frames = 200
v      = 1
w      = 6 * np.pi
k      = w / v

fig, ax = plt.subplots(1, 1, figsize = (10,6))
plt.rc('font', size = 16)
plt.close()

x = np.linspace(0, 1, 100)
t = np.linspace(0, 2, frames)

ax.set_ylim(-2.1, 2.1)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid()

ln0, = ax.plot(x, + np.cos(k *x - w * t[0]), '--b', lw = 3)
ln1, = ax.plot(x, - np.cos(k *x + w * t[0]), '--g', lw = 3)
ln2, = ax.plot(x, np.cos(k *x - w * t[0]) - np.cos(k *x + w * t[0]), '-r', lw = 3)

def update(i):
    ln0.set_data(x, + np.cos(k *x - w * t[i]) )
    ln1.set_data(x, - np.cos(k *x + w * t[i]) )
    ln2.set_data(x, np.cos(k *x - w * t[i]) - np.cos(k *x + w * t[i]) )

plt.tight_layout()
animation_fig = animation.FuncAnimation(fig, update, frames = frames, interval = 50)
#HTML(animation_fig.to_jshtml())
animation_fig.save("ANIM03_StandingWave.mp4", dpi = 200)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

frames = 100
v      = 1
w0     = np.pi
k0      = w0 / v

fig, axes = plt.subplots(2, 2, figsize = (10,6))
plt.rc('font', size = 16)
plt.close()

x = np.linspace(0, 1, 100)
t = np.linspace(0, 2, frames)

def update(i):
    for n, ax in enumerate(axes.ravel(), 1):
        ax.cla()
        ax.set_ylim(-1.1, 1.1)
        ax.grid()
        ax.plot(x, np.sin(n * k0 * x) * np.cos(n * w0 * t[i]), '-b', lw = 3)
        ax.plot([0,1], [0,0], 'ob')

plt.tight_layout()
animation_fig = animation.FuncAnimation(fig, update, frames = frames, interval = 50)
#HTML(animation_fig.to_jshtml())
animation_fig.save("ANIM04_FixedFixedModes.mp4", dpi = 200)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
from IPython.display import HTML

theta = np.linspace(0, 2 * np.pi, 100)

fig, ax = plt.subplots(figsize = (12, 6.2))
plt.rc('font', size = 16)
plt.close()

def update(i):

    vel = 0.1 * i
    
    ax.cla()
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_facecolor('cyan')
    ax.set_xlim(-3.0, 1.1)
    ax.set_ylim(-1.1, 1.1)
    ax.fill([-0.25, -0.25, 0.1], [0.02, -0.02, 0], color = 'k')
    ax.fill([-0.15, -0.15, -0.10], [0.2, -0.2, 0], color = 'k')
    ax.fill([-0.25, -0.25, -0.2], [0.05, -0.05, 0], color = 'k')
    fronts = [ax.plot( r * np.cos(theta) - vel * r, r * np.sin(theta) , 'b', lw = 0.5) for r in np.arange(0, 2, 0.1)]
    if vel > 1:
        mach = ax.plot([-3, 0, -3], [3 * np.tan(np.arcsin(1/vel)), 0, -3 * np.tan(np.arcsin(1/vel))], '--k')
    ax.arrow(0, 0, vel/2, 0, color = 'red', width = .01, zorder=3)
    ax.text(0.3, 0.1, f'v = {vel:.2f} v$_s$', fontdict = {'size': 12, 'color': 'red'})

    return fig

plt.tight_layout()
animation_fig = animation.FuncAnimation(fig, update, frames = 20, interval = 500)
#HTML(animation_fig.to_jshtml())
animation_fig.save("ANIM07_DopplerEffect.mp4", dpi = 200)
```
